# Number gussing game Level 1

## Print if the user's guessed number is right or not.

```python
secret_number = 55
user_input = int(input("enter your reference number"))
if secret_number == user_input:
    print("congratualations you find the number")
else:
    print("sorry your fail to find")
```

# Number guessing game Level 2

### 2.1 Using a For Loop  The end iteration number is defined.

```python
secret_number = 55
for i in range(0,7):
    user_input = int(input("enter your reference number"))
    if secret_number == user_input:
        print("congratualations you find the number")
        break
    else:
        print("sorry your fail to find")
```

## 2.2 While Loop- when end number is not defind and the insturctions has to excuted infiite times

```python
secret_number = 55

while True:
    user_input = int(input("enter your reference number"))
    if secret_number == user_input:
        print("congratualations you find the number")
        break
    else:
        print("sorry your fail to find")
```

# Number guessing game Level 3 

## Give multiple conditions

1. The user should enter the number between 1 and 100. If it's entered wrong, the program should say, "invalid number."
2. If the user's guessed number > secret number, then the program should say, "Your number is greater than the number."
3. If the user's guessed number is less than the secret number, then the program should say, "Your number is less than the secret number."

```python
print("Welcome to Number Gussing Game!!!")
print("Gussing the Number Between 1-100")

secret_number = 55
for i in range(0,7):
    user_input = int(input("enter your reference number"))
    if user_input>100 or user_input < 1:
        print("Invalid number! please enter the number between 1-100")
    elif user_input > secret_number:
        print("your number higher")
    elif user_input < secret_number:
        print("your number is lesser")  
    else:
        print("your correct number")
        print(f"congratulations, you have gussed the number {secret_number} in {i} attempts")
        break
```
